# Sentiment Analysis with Transformers

This notebook demonstrates sentiment analysis using pre-trained transformer models from Hugging Face.

## What is Sentiment Analysis?

Sentiment analysis is the process of determining the emotional tone behind text. It's commonly used to understand opinions, emotions, and attitudes expressed in text data.

## Setup

First, let's import the necessary libraries:

```python
from transformers import pipeline
import pandas as pd
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

## Load Pre-trained Model

We'll use DistilBERT, a lighter and faster version of BERT, fine-tuned on the SST-2 sentiment classification dataset.

```python
# Load the sentiment analysis pipeline
sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("Model loaded successfully!")
```

## Single Text Analysis

Let's analyze the sentiment of a single text:

```python
# Example text
text = "I absolutely love this product! It's amazing!"

# Analyze sentiment
result = sentiment_analyzer(text)[0]

print(f"Text: {text}")
print(f"Sentiment: {result['label']}")
print(f"Confidence: {result['score']:.4f}")
```

## Batch Analysis

Now let's analyze multiple texts at once:

```python
# Multiple example texts
texts = [
    "I absolutely love this product! It's amazing!",
    "This is the worst experience I've ever had.",
    "The movie was quite entertaining and well-made.",
    "I'm disappointed with the quality and service.",
    "This book changed my life! Highly recommended!",
    "Terrible customer support, very frustrating.",
    "The weather is nice today.",
    "I'm not sure how I feel about this.",
]

# Analyze all texts
results = sentiment_analyzer(texts)

# Create a DataFrame for better visualization
df = pd.DataFrame({
    'Text': texts,
    'Sentiment': [r['label'] for r in results],
    'Confidence': [r['score'] for r in results]
})

print(df.to_string(index=False))
```

## Visualize Results

Let's create a visualization of the sentiment distribution:

```python
# Count sentiments
sentiment_counts = df['Sentiment'].value_counts()

# Create bar plot
plt.figure(figsize=(10, 6))
sentiment_counts.plot(kind='bar', color=['green', 'red'])
plt.title('Sentiment Distribution', fontsize=16, fontweight='bold')
plt.xlabel('Sentiment', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Average confidence by sentiment
avg_confidence = df.groupby('Sentiment')['Confidence'].mean()
print(f"\nAverage Confidence by Sentiment:")
print(avg_confidence)
```

## Interactive: Try Your Own Text!

Modify the text below to analyze your own sentences:

```python
# Try your own text here!
your_text = "Enter your text here to analyze its sentiment."

result = sentiment_analyzer(your_text)[0]

print(f"Your text: {your_text}")
print(f"Predicted sentiment: {result['label']}")
print(f"Confidence: {result['score']:.4f}")

# Visual representation
fig, ax = plt.subplots(figsize=(8, 2))
colors = ['green' if result['label'] == 'POSITIVE' else 'red']
ax.barh(['Sentiment'], [result['score']], color=colors)
ax.set_xlim(0, 1)
ax.set_xlabel('Confidence Score')
ax.set_title(f"Sentiment: {result['label']} ({result['score']:.2%})")
plt.tight_layout()
plt.show()
```
